```python
paragraph = "He is a good boy. She is a good girl. They are good students."
```

#### Data Preprocessing
##### Tokenization
**Sentence Tokenization** : breaking into sentences

```python
sentences = nltk.sent_tokenize(paragraph)
print(sentences)
```

**Word Tokenization** : breaking into words

```python
words = nltk.word_tokenize(paragraph)
print(words)
```

**List Stop Words**

```python
from nltk.corpus import stopwords

# Remove stop words : Here we are removing common words that do not add much meaning 
# to the text, such as "is", "a", "the", etc.
stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word not in stop_words]
print(filtered_words)

# stopwords.words("english") will return a list of common stop words in English language. 
```

**Stemming** : Bring the given word to the base word

```python
## Stemming : Stemming is the process of reducing a word to its base or root form.
## For example, the words "running", "runner", and "ran" can all be reduced to the stem "run".
## Applications of stemming include:
## 1. Information Retrieval: Stemming can improve search results by allowing users to find relevant
##    documents even if they use different forms of a word. For example, a search for "run" would also return results containing "running" and "runner".
## 2. Text Classification: Stemming can help in text classification tasks by reducing the dimensionality of the feature space.
##   For example, in sentiment analysis, stemming can help to group similar words together, improving the performance of the classifier.

## No human intervention is required in stemming as it is an automated process that can be applied to large volumes of text data without the need for manual input. 
# Stemming algorithms, such as the Porter Stemmer, are designed to handle various word forms and can be applied consistently across different texts.

from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print(stemmed_words)
```

**Lemmatization** : Bring the given word to the base word

```python
## Lemmatization : Lemmatization is similar to stemming, 
## but it reduces a word to its base or dictionary form, called a lemma.
## For example, the words "running", "runner", and "ran" can all be reduced to the lemma "run".
## Different from stemming, lemmatization considers the context and the meaning of the word,
## and it returns a valid word that exists in the language.
## For example, the word "better" can be lemmatized to "good", while stemming would return "better".
## Applications of lemmatization include:
## 1. Natural Language Processing: Lemmatization can improve the performance of NLP tasks by
##    reducing words to their base forms, which can help in tasks like part-of-speech tagging, named entity recognition, and machine translation.
## 2. Text Analysis: Lemmatization can help in text analysis tasks by grouping similar
##    words together, which can improve the accuracy of sentiment analysis, topic modeling, and other text analysis techniques.

## Human intervention is required in lemmatization to some extent, 
## as it may require additional information about the context of the word to determine the correct lemma.

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print(lemmatized_words)
```

#### Data Cleaning
1. **Remove punctuation**
2. **Remove stopwords**
3. **Stemming / Lemmatiation**

```python
import pandas as pd
df = pd.DataFrame({
    "in" : ["He is a good boy.", "She is a good girl.", "They are good students."],
    "out" : ["P" , "P", "N"]
})
print(df)
```

```python
## Now we will apply the same preprocessing steps to the text data in the DataFrame. 
# We will create a new column called "corpus" that contains the preprocessed text data.

corpus = []
import re
for i in range(len(df)):
    rp = re.sub("[^a-zA-Z]", " ", df["in"][i])
    rp = rp.lower()
    rp = rp.split()
    rp = [stemmer.stem(word) for word in rp if word not in stop_words]
    rp = " ".join(rp)
    corpus.append(rp)
print(corpus)

## Above code snippet is performing the following steps:
## 1. It initializes an empty list called "corpus" to store the preprocessed text data.
## 2. It uses a for loop to iterate through each row of the DataFrame "df".
## 3. Inside the loop, it uses the re.sub() function to remove any non-alphabetic characters from the text in the "in" column of the DataFrame.
## 4. It converts the text to lowercase using the lower() method.
## 5. It splits the text into individual words using the split() method.
## 6. It applies stemming to each word in the text using a list comprehension and the Porter Stemmer.
## 7. It joins the stemmed words back into a single string using the join() method and appends it to the "corpus" list.
```
