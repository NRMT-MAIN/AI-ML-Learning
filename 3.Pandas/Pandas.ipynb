### Pandas Series

- It is a one-dimensional (1D) array holding data of any type.
- A Pandas Series is like a column in a table.

> **Note:**
> 1. dtype is important
> 2. Unlike NumPy, data will not be converted according to maximum data type.
> 3. Unlike NumPy, error will not be thrown when in heterogenous shape
>
> `ndim`, `shape` etc. are same as NumPy
>
> In Numpy, only list is used to convert to array. In pandas, both list and dictionary are used to convert to series.
>
> In pandas we can provide index other than number — understand by `iloc` and `loc`.
>
> For patterns like arange, linspace you can use numpy and then convert to series.

#### Combining DataFrames
##### Merging

```python
df1 = pd.DataFrame({
    "city": ["mumbai", "delhi", "banglore", "hyderabad"],
    "temperature": [32, 45, 40, 36]
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["delhi", "mumbai", "banglore", "chennai"],
    "humidity": [68, 65, 75, 70]
})
df2
```

```python
df3 = pd.read_excel(r"file_path.xlsx") 
# or use forward slash to read
# This will read the Excel file located at "file_path.xlsx" and 
# create a DataFrame called df3 with the data from the Excel file.
```

```python
# Merging the DataFrames based on the common column "city"
pd.merge(df1, df2, on="city", how="inner")
# This will merge df1 and df2 based on the common column "city".
# The "on" parameter specifies the column to merge on, and the "how" parameter specifies the type of merge to perform. In this case, "inner" means that only the rows with matching values in the "city" column will be included in the merged DataFrame.
```

```python
pd.merge(df1, df2, on="city", how="outer")
# This will merge df1 and df2 based on the common column "city".
# The "how" parameter specifies the type of merge to perform. In this case, "outer" means that all rows from both DataFrames will be included in the merged DataFrame, with NaN values filling in for any missing data.
```

```python
pd.merge(df1, df2, on="city", how="right")
# This will merge df1 and df2 based on the common column "city".
# The "how" parameter specifies the type of merge to perform. In this case, "right" means that all rows from df2 will be included in the merged DataFrame, and only the matching rows from df1 will be included. If there are any missing values in df1 for the matching rows, they will be filled with NaN.
```

##### Concatenation

```python
# In concat you can combine multiple DataFrames along a particular axis (rows or columns) without the need for a common column.
pd.concat([df1, df2], axis=1)
# This will concatenate df1 and df2 along the columns (axis=1). 
# The resulting DataFrame will have all the columns from both df1 and df2, and the rows will be aligned based on their index. 
# If there are any missing values in either DataFrame for the aligned rows, they will be filled with NaN.
```
