**Statistical Measures**
- Mean = average --> df["col"].mean()
- Median = centre position value of a ordered data --> df["col"].median
No. of Data points below median value = No. of Data points above median value
- Mode = most repeated value df["col"].mode()[0]

Perccnetage (Comparision with other) : 
Percentile (No comparision with other): no. of datapoints below to value / total no. of values.

Salaries 
 0 percentile = min value
 25 percentile = Q1
 50 percentile = median Q2
 75 percentile = Q3
 100 percentile = max value

Inter Quartile Range (IQR) : Q3 - Q1
lower limit = Q1 - 1.5 * IQR
upper limit = Q3 + 1.5 * IQR

Outliers/Abnormal Values : values which are very far from the other values in the data set.

##### Outliers

```python
df= pd.DataFrame({"marks": [10, 11,12, 25, 25, 27, 31, 33, 34, 34, 36, 36, 43, 50, 59]})
df
```

Various ways of finding the outliers
1. Boxplot - Identifying Outliers based on boxplot

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(x=df["marks"])
plt.show()
```

```python
#calculate Q1
Q1=df["marks"].quantile(0.25)
print("Q1:",Q1)
# Q1 is the value below which 25% of the data falls. It is also known as the first quartile. 
# In this case, Q1 is 25, which means that 25% of the marks are less than or equal to 25.

#calculate Q3
Q3=df["marks"].quantile(0.75)
print("Q3:",Q3)
# Q3 is the value below which 75% of the data falls. It is also known as the third quartile. 
# In this case, Q3 is 36, which means that 75% of the marks are less than or equal to 36.

#calculate IQR
IQR = Q3 - Q1
print("IQR:",IQR)
# IQR is the interquartile range, which is the difference between Q3 and Q1. 
# It represents the spread of the middle 50% of the data.

#Calculate lower limit of outlier
lower_limit = Q1 - (IQR * 1.5)
print("lower limit:",lower_limit)

#Calculate upper limit of outlier
upper_limit = Q3 + (IQR * 1.5)
print("upper limit:", upper_limit)
```

```python
df[(df["marks"] < lower_limit) | (df["marks"] > upper_limit)]
```
