Types of Columns
- Numerical Columns: These columns contain numerical data, such as integers or floats. Examples include "total_bill", "tip", and "size".
- Categorical Columns: These columns contain categorical data, which can be either nominal (no inherent
order) or ordinal (inherent order)

Structured Data Types
1. Cross-Sectional Data
Definition: Non-sequential data where the order of rows is not important.
2. Time-Series Data
Definition: Sequential data where the order of rows matters.

Numerical Data (Quantitative)
1. Continuous
Columns with float values.
Columns with only integer values can also be considered continuous (or counts).
2. Discrete Count
Represents individual values (whole numbers).

Text Data (Qualitative)
Discrete Categorical
1. Nominal
Definition: Categories with no natural order.
Examples:
Letters: A, B, C, D
Names: Pavan, Prabhas, Kalyan, Ram
Countries: India, US, Russia
2. Ordinal
Definition: Categories with a natural sequence.
Examples:
Letter grades: A, B, C, D
Shirt sizes: S, M, L, XL, XXL

#### Data Preprocessing or Data Preparation
- Identify
- Extract
- Treatment

Data Cleaning
   1. Wrong Data : Identify + Extract/Seperate + Treatment(Replace ore Remove)
   2. Wrong Data Type : Identify + Extract/Seperate + Treatment(Convert)
   3. Duplicates : Identify + Extract/Seperate + Treatment(remove)
   4. Missing Values : Identify + Extract/Seperate + Treatment(replace ore remove)
   5. Abnormal Values : Identify + Extract/Seperate + Treatment(Replace ore Remove or retrain)
   6. Unimportant Variables : Identify + Extract/Seperate + Treatment (remove)


In entire data cleaning , always 1st or best option : replace with original Data
If 1st option is not avaialable then 2nd option : replace with mean/median/mode
