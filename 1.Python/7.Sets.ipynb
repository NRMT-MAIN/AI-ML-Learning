#### Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for
membership tests, eliminating duplicate entries, and performing mathematical set operations like union,intersection, difference, and symmetric difference.

```python
## create a set
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(my_empty_set)
```

```python
my_empty_set = set([1, 2, 3])
print(my_empty_set)
```

```python
## Basics Set Operations
## Adding and removing elements from a set

my_set.add(6)
print(my_set)
my_set.remove(3)
print(my_set)

## If we add the same element again, it will not be added to the set -> ignore
```

```python
my_set.discard(11)  # discard will not raise an error if the element is not present
print(my_set)
```

```python
# pop() method removes and returns an arbitrary element from the set. 
# If the set is empty, it raises a KeyError.
removed_element = my_set.pop()
print(f"Removed element: {removed_element}")
```

```python
## clear all the elements from the set
my_set.clear()
print(my_set)
```

```python
## Set Membership Testing
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # True
print(6 in my_set)  # False
```

```python
## Mathematical Set Operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
# Union
union_set = set_a.union(set_b)
print(f"Union: {union_set}")

# Intersection
intersection_set = set_a.intersection(set_b)
print(f"Intersection: {intersection_set}")

# Difference
difference_set = set_a.difference(set_b)
print(f"Difference (A - B): {difference_set}")

# Symmetric Difference
sym_diff_set = set_a.symmetric_difference(set_b)
print(f"Symmetric Difference: {sym_diff_set}")
```

```python
## Sets Method
# issubset() method checks if the set is a subset of another set
set_c = {1, 2}
print(set_c.issubset(set_a))  # True
print(set_a.issubset(set_c))  # False

# issuperset() method checks if the set is a superset of another set
print(set_a.issuperset(set_c))  # True
print(set_c.issuperset(set_a))  # False

# isdisjoint() method checks if two sets have no elements in common
set_d = {5, 6}
print(set_a.isdisjoint(set_d))  # True
print(set_a.isdisjoint(set_b))  # False
```

```python
## Couting Unique words in text
text = "hello world hello"
unique_words = set(text.split())
print(unique_words)
```
