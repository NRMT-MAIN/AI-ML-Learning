#### Tuples
Tuples are ordered, immutable collections of items. They are defined using parentheses `()` 
and can contain elements of different types. Once a tuple is created, its contents cannot be changed.

```python
## creating a tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list(range(5))
tup = tuple(lst)
print(tup)
```

```python
mixed_tuple = (1, "hello", 3.14, [1, 2, 3], (4, 5))
print(mixed_tuple)
```

```python
## Accessing tuple elements
print(mixed_tuple[0])  # 1
print(mixed_tuple[1])  # "hello"
print(mixed_tuple[3])  # [1, 2, 3]
print(mixed_tuple[4])  # (4, 5)
```

```python
print(mixed_tuple[-1])  # (4, 5)
print(mixed_tuple[-2])  # [1, 2, 3]
print(mixed_tuple[0:3])  # (1, "hello", 3.14)
```

```python
# Tuple Operations
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated_tuple = tuple1 + tuple2
print(concatenated_tuple)  # (1, 2, 3, 4, 5, 6)

repeated_tuple = tuple1 * 3
print(repeated_tuple)  # (1, 2, 3, 1, 2, 3, 1, 2, 3)
```

```python
# Immutable Nature of Tuples
# Tuples are immutable, which means you cannot change their contents after creation.
# The following code will raise a TypeError:
try:
    mixed_tuple[0] = 10
except TypeError as e:
    print(e)  # 'tuple' object does not support item assignment
```

```python
## Tuple Methods
# Tuples have only two built-in methods: count() and index()
sample_tuple = (1, 2, 3, 2, 4, 2)
print(sample_tuple.count(2))  # 3

print(sample_tuple.index(3))  # 2
```

```python
# Packing and Unpacking
# Tuple packing allows you to create a tuple without using parentheses.
packed_tuple = 1, "hello", 3.14
print(packed_tuple)  # (1, "hello", 3.14)

# Tuple unpacking allows you to assign the elements of a tuple to variables in a single statement.
a, b, c = packed_tuple
print(a)  # 1
print(b)  # "hello"
print(c)  # 3.14
```

```python
# Unpacking with asterisk (*)
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
print(first)   # 1
print(middle)  # [2, 3, 4]
print(last)    # 5
```

```python
# Nested tuples
nested_tuple = (1, (2, 3), (4, 5, 6))
print(nested_tuple)  # (1, (2, 3), (4, 5, 6))
print(nested_tuple[1])  # (2, 3)
```
