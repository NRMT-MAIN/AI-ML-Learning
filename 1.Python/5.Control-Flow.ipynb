#### Conditional_Statements

```python
## If statement
x = 10
if x > 5:
    print("x is greater than 5")

## If-else statement
y = 3
if y > 5:
    print("y is greater than 5")
else:    
    print("y is not greater than 5")

## If-elif-else statement
z = 7
if z > 10:
    print("z is greater than 10")
elif z > 5:
    print("z is greater than 5 but not greater than 10")
else:
    print("z is not greater than 5")
```

```python
## Nested Conditional Statements
a = 15
if a > 10:
    if a % 2 == 0:
        print("a is greater than 10 and even")
    else:
        print("a is greater than 10 and odd")
else:
    print("a is not greater than 10")
```

```python
## Practice Exercise
## Determine if a year is a leap year using nested conditional statements

year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

#### Loops

```python
## For Loops
for i in range(5):
    print(i)

print("------------------")

for i in range(1, 6, 2):
    print(i)

print("------------------")
for i in range(10, 0, -2):
    print(i)
```

```python
## While Loops
count = 0
while count < 5:
    print(count)
    count += 1
print("------------------")
number = 10
while number > 0:
    print(number)
    number -= 2
print("------------------")
```

```python
## Break and Continue Statements
for i in range(10):
    if i == 5:
        break
    print(i)
print("------------------")
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
```

```python
## Pass Statement - used as a placeholder for future code
for i in range(5):
    if i == 2:
        pass  # This will be replaced with actual code later
    print(i)
```

```python
## Nested Loops
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```
