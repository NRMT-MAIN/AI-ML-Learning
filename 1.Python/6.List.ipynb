```python
## Example 1 : Manage a TODO list

todo_list = ["Buy groceries", "Clean the house", "Pay bills"]

## Adding to task
todo_list.append("Call mom")
print(todo_list)

## Removing a task
todo_list.remove("Clean the house")
print(todo_list)

## Checking if a task is in the list
if "Pay bills" in todo_list:
    print("You have to pay bills.")

## Iterating through the list
print("Your TODO list:")
for task in todo_list:
    print("- " + task)
```

```python
## Example 2 : Organizing student grades
grades = [85, 92, 78, 90, 88]

## Adding a new grade
grades.append(95)

## Calculating the average grade
average_grade = sum(grades) / len(grades)
print(f"Average grade: {average_grade:.2f}")

## Finding the highest and lowest grades
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Highest grade: {highest_grade}")
print(f"Lowest grade: {lowest_grade}")
```

```python
## Example 3 : Managing a inventory
inventory = ["Apples", "Bananas", "Oranges"]

## Adding new items to the inventory
inventory.append("Grapes")
print(inventory)

## Removing an item from the inventory
inventory.remove("Bananas")

print(inventory)

## Checking if an item is in the inventory
if "Oranges" in inventory:
    print("Oranges are in stock.")

## Iterating through the inventory
print("Current inventory:")
for item in inventory:
    print("- " + item)
```

```python
## Example 4 : Collecting User feedback
feedback_list = ["Great service!", "Could be better.", "Loved the experience!"]

## Adding new feedback
feedback_list.append("Not satisfied with the product.")

## Counting specific feedback
positive_feedback_count = sum(1 for feedback in feedback_list if "Great" in feedback or "Loved" in feedback)
print(f"Number of positive feedback: {positive_feedback_count}")

## Iterating through feedback
print("User feedback:")
for feedback in feedback_list:
    print("- " + feedback)
```
