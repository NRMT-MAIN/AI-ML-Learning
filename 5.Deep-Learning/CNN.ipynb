#### Step 1 : Business Problem Understanding
- Classify the image
- Predict whether a given image is cat or dog

```python
import tensorflow as tf 
import keras  
```

#### Step 2 : Load the image data & Image Data Preprocessing 

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_datagen = ImageDataGenerator(rescale = 1/255)
# Above code is used to rescale the pixel values of the images to be between 0 and 1. 
# This is done by dividing each pixel value by 255, which is the maximum pixel 
# value for an 8-bit image. Rescaling helps in improving the performance of 
# the neural network by normalizing the input data.
```

```python
training_set = train_datagen.flow_from_directory("D:\\Study_Material\\AI\\dataset\\dataset\\training_set",
                                                 target_size = (64, 64),
                                                 class_mode = 'binary')
# Above code is used to create a training set from the images in the specified directory. 
# The flow_from_directory method is used to load the images from the directory 
# and apply the rescaling transformation defined in the train_datagen.
# The target_size parameter is used to resize the images to a specific size (64x64 in this case).
# The class_mode parameter is set to 'binary' because we are performing binary classification (cat vs dog).
```

```python
training_set.class_indices
```

```python
test_datagen = ImageDataGenerator(rescale = 1/255)
# Similar to the training set, we also need to rescale the pixel values of the test images.

test_set = test_datagen.flow_from_directory("D:\\Study_Material\\AI\\dataset\\dataset\\test_set",
                                            target_size = (64, 64),
                                            class_mode = 'binary')
# Multiple images are loaded from the specified directory for the test set, and the same rescaling transformation is applied.
# Similar to the training set, we create a test set from the images in the specified directory.
```

#### Modelling - Convolution Neural Network
**Initialising the CNN**

```python
from keras.models import Sequential
classifier = Sequential()
```

**Step 1 - Convolution**

```python
from keras.layers import Conv2D

classifier.add(Conv2D(input_shape = [64, 64, 3], # for RGB images, we have 3 channels (Red, Green, Blue)
                      filters = 32, # max filters = 32
                      kernel_size = 3, # kernelsize = filter size
                      activation = 'relu'))
```

**Step 2 - Max Pooling**

```python
from keras.layers import MaxPooling2D

classifier.add(MaxPooling2D(pool_size = 2, # pool size is the size of the window that will be used to perform max pooling
                            strides = 2)) # strides is the number of pixels that the window will move after each pooling operation
```

**Step 3 - Flattening**

```python
from keras.layers import Flatten
classifier.add(Flatten())
```

**Step 4 - Full Connections**

```python
from keras.layers import Dense

# hidden layer with 128 neurons
classifier.add(Dense(units = 128, activation = 'relu'))
## 128 beacuse it's a common practice to use a power of 2 for the number of neurons in a layer, 
##  and 128 (64 * 2) is a good choice for a hidden layer in a CNN.

# output layer with 1 neuron (since it's binary classification)
classifier.add(Dense(units = 1, activation = 'sigmoid'))
# For multi-class classification, we would use 'softmax' activation and set units to the number of classes.
```

**Training the CNN Model with train data & Testing the model with test data**

```python
classifier.compile(optimizer = 'adam' , 
                   loss = 'binary_crossentropy', 
                   metrics = ['accuracy'])
# multiclass : loss = 'categorical_crossentropy'
```

```python
classifier.fit(x = training_set,
               validation_data = test_set,
                epochs = 25)
```

#### Evaluation
**Making a single predictions**
