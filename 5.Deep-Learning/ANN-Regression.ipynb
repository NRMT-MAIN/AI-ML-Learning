**Input and Ouput Variables**

```python
x = df.drop("PE", axis=1)
y = df["PE"]
```

**Splitting the data into training and testing data**

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
# above code will split the data into 80% training data and 20% testing data. random_state is used to get the same split every time we run the code.
# random_state is used to get the same split every time we run the code. if we do not set random_state then we will get different split every time we run the code.
# we can set any number for random_state. 
```

##### Perceptron

```python
import tensorflow as tf
import keras
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

**Design the perceptron model**

```python
## initialize of neural network
model = Sequential()

## add input layer and first hidden layer
model.add(Dense(input_dim=4, units=1, activation="relu", kernel_initializer="uniform")) 
# input_dim is number of features in the data. units is number of neurons in the hidden layer. activation is activation function used in the hidden layer.
# kernel_initializer is used to initialize the weights of the neurons. we can use uniform or normal distribution to initialize the weights.
# dense connects all the neurons in the previous layer to all the neurons in the next layer.

# compile - calculate the loss and optimizer
model.compile(loss="mean_squared_error", optimizer="adam")
```

```python
# fit the train data to the model for calculating the weights and bias of the neurons.
model.fit(x_train, y_train, epochs=100, batch_size=32)
```

```python
### Prediction on train data
y_train_pred = model.predict(x_train)
y_train_pred

## evaluate the train data
from sklearn.metrics import mean_squared_error
mse_train = mean_squared_error(y_train, y_train_pred)
print("Mean Squared Error on Train Data: ", mse_train)
print("Train RMSE: ", np.sqrt(mse_train))
```

#### ANN Model
- design
- fit
- predict
- evaluation (RMSE)

```python
# initoalization of neural network
nn1 = Sequential()

# add input layer and hidden layer
nn1.add(Dense(input_dim=4, units=8, activation="relu", kernel_initializer="uniform"))
nn1.add(Dense(units=1, activation="relu", kernel_initializer="uniform"))

# compile the model
nn1.compile(loss="mean_squared_error", optimizer="adam")
```

```python
nn1.fit(x_train, y_train, epochs=100, batch_size=32)
```

```python
## Prediction on train data
y_train_pred = nn1.predict(x_train)
```

```python
## evaluate the train data
from sklearn.metrics import mean_squared_error
mse_train = mean_squared_error(y_train, y_train_pred)
print("Mean Squared Error on Train Data: ", mse_train)
print("Train RMSE: ", np.sqrt(mse_train))
```

**With 2 hidden layer**

```python
nn2 = Sequential()
nn2.add(Dense(input_dim=4, units=8, activation="relu", kernel_initializer="uniform"))
nn2.add(Dense(units=8, activation="relu", kernel_initializer="uniform"))
nn2.add(Dense(units=1, activation="relu", kernel_initializer="uniform"))
nn2.compile(loss="mean_squared_error", optimizer="adam")
nn2.fit(x_train, y_train, epochs=100, batch_size=32)
y_train_pred = nn2.predict(x_train)
mse_train = mean_squared_error(y_train, y_train_pred)
print("Mean Squared Error on Train Data: ", mse_train)
print("Train RMSE: ", np.sqrt(mse_train))
```

Once all epochs are completed , after last epoch : weights will be stored and used for future prediction.

Train RMSE == Train RMSE (good model) +-5% : Here we can say that our model is good because train RMSE and test RMSE are almost same.

Train RMSE > Test RMSE (underfitting problem) : Here we can say that our model is underfitting because train RMSE is greater than test RMSE. it means that our model is not able to capture the underlying pattern of the data.
- Model not performed well on trained data
- Model performed well on tested data
- Solution : Add more amount of data (add more records)

Train RMSE < Test RMSE (overfitting problem) : Here we can say that our model is overfitting because train RMSE is less than test RMSE. it means that our model is too complex and is memorizing the training data instead of learning the underlying pattern.
- Model performed well on trained data
- Model not performed well on tested data
- Solution : depends on algorithm

**Improve the accuracy in ANN**
1. Valid data cleaning
2. increase epochs
3. increase hidden layers
4. increase batch_size
