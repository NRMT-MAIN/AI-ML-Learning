**Splitting the data into training and testing data**

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
# above code will split the data into 80% training data and 20% testing data. random_state is used to get the same split every time we run the code.
# random_state is used to get the same split every time we run the code. if we do not set random_state then we will get different split every time we run the code.
# we can set any number for random_state. 
```

##### Perceptron

```python
import tensorflow as tf
import keras
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

**Design the perceptron model**

```python
## initialize of neural network
model = Sequential()

## add input layer and first hidden layer
model.add(Dense(input_dim=4, units=1, activation="relu", kernel_initializer="uniform")) 
# input_dim is number of features in the data. units is number of neurons in the hidden layer. activation is activation function used in the hidden layer.
# kernel_initializer is used to initialize the weights of the neurons. we can use uniform or normal distribution to initialize the weights.
# dense connects all the neurons in the previous layer to all the neurons in the next layer.

# compile - calculate the loss and optimizer
model.compile(loss="mean_squared_error", optimizer="adam")
```

```python
# fit the train data to the model for calculating the weights and bias of the neurons.
model.fit(x_train, y_train, epochs=100, batch_size=32)
```

```python
### Prediction on train data
y_train_pred = model.predict(x_train)
y_train_pred

## evaulate the 
```
