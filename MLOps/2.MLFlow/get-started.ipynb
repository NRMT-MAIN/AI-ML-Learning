#### MLFlow Tracking Server
Here are the steps to set up an MLFlow Tracking Server:
1. **Install MLFlow**: You can install MLFlow using pip:
```bash
pip install mlflow
```
2. **Start the MLFlow Tracking Server**: You can start the MLFlow Tracking Server using the following command:
```bash
mlflow server --backend-store-uri sqlite:///mlflow.db --default-artifact-root
file:///mlflow-artifacts
```
This command will start the MLFlow Tracking Server with a SQLite database as the backend store and a local directory for artifact storage.  

3. **Access the MLFlow UI**: You can access the MLFlow UI by navigating to 
```bash
mlflow ui
```
`http://localhost:5000` in your web browser. Here, you can view and manage your MLFlow experiments, runs, and artifacts.

4. **Log Experiments**: You can log experiments to the MLFlow Tracking Server using the MLFlow Python API. Here’s an example of how to log a simple experiment:

```python
import mlflow
with mlflow.start_run():
    mlflow.log_param("param1", 5)
    mlflow.log_metric("metric1", 0.85)
    mlflow.log_artifact("path/to/artifact")
```
This code will log a parameter, a metric, and an artifact to the MLFlow Tracking Server under the current run.
5. **View Logged Experiments**: You can view the logged experiments, parameters, metrics, and artifacts in the MLFlow UI. You can compare different runs, visualize metrics, and manage your experiments effectively.
Make sure to replace `path/to/artifact` with the actual path to your artifact file.
By following these steps, you can set up and use the MLFlow Tracking Server to manage your machine learning experiments effectively.

````python
import mlflow
````

````python
mlflow.set_tracking_uri("http://localhost:5000")
# Above code sets the tracking URI to the local MLflow server running on port 5000. 
# This allows you to log experiments and metrics to your local MLflow server.
````

````python
mlflow.set_experiment("Check localhost connection")
# Above code sets the current experiment to "Check localhost connection". 
# If the experiment does not exist, it will be created.

with mlflow.start_run():
    mlflow.log_param("param1", 1)
    mlflow.log_metric("test", 1)
    mlflow.log_metric("nirmit", 2)
# Above code starts a new MLflow run and logs a parameter "param1" with value 1, and two metrics "test" with value 1 and "nirmit" with value 2.
````
